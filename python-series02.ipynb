### Plot sinusoidal signals

The goal of this notebook is to code something useful from therory to code regarding sinusoidal signals. The goal is to plot sinusoidal signal. 

## Maths to generate a sine wave 

The equation of a sine wave is:

$$ 
y(t)= A\sin(\omega t + \phi)
$$

where:
- $A$ is the amplitude (unit of the physical system we simulate)
- $\omega$ is the angular frequency (rad/s)
- $\phi$ is the phase (rad)

$ : It is latex language to write maths in a nice way 

```python
# import library 

import numpy as np 
import matplotlib.pyplot as plt

# Create a plot 

# Create a numpy array called t that contains values from 0 to 1 
t = np.linspace(0,1,1000)
# Create a numpy array called s that contains the values of a sine fonction of t (s is for signal)
s = np.sin(2* np.pi * 5 * t + np.pi/4) # 2pi : one wave 
# Plot s as a function of t 
plt.plot(t,s)
# Add a title to the plot
plt.title('Sinusoidal function')
# Add a label to the x axis
plt.xlabel('Time (s)')
# Add a label to the y axis
plt.ylabel('Amplitude')
# Add a grid to the plot
plt.grid(True)
# Save the plot as a png file
plt.savefig('sine_wave_png')
```
