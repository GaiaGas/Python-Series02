# Plot sinusoidal signals

The goal of this notebook is to code something useful from therory to code regarding sinusoidal signals. The goal is to plot sinusoidal signal. 

## Maths to generate a sine wave 

The equation of a sine wave is:

$$ 
y(t)= A\sin(\omega t + \phi)
$$

where:
- $A$ is the amplitude (unit of the physical system we simulate)
- $\omega$ is the angular frequency (rad/s)
- $\phi$ is the phase (rad)

$ : It is latex language to write maths in a nice way 

## First draft 

```python
# import library 

import numpy as np 
import matplotlib.pyplot as plt

# Create a plot 

# Create a numpy array called t that contains values from 0 to 1 
t = np.linspace(0,1,1000)
# Create a numpy array called s that contains the values of a sine fonction of t (s is for signal)
s = np.sin(2* np.pi * 1 * t + 3*np.pi/4) # 2pi : one wave, add a phase dalay as in the formula, could be an other value it is just an example.
# Plot s as a function of t 
plt.plot(t,s)
# Add a title to the plot
plt.title('Sinusoidal function')
# Add a label to the x axis
plt.xlabel('Time (s)')
# Add a label to the y axis
plt.ylabel('Amplitude')
# Add a grid to the plot
plt.grid(True)
# Save the plot as a png file
plt.savefig('sine_wave_png')
```

## Better version


```python
# Create a plot with sine and cosine

# Create a numpy array called t that contains values from 0 to 1 
t = np.linspace(0,1,1000)
# Create a numpy array called s that contains the values of a sine fonction of t (s is for signal)
s = np.sin(2* np.pi * t) # 2pi : one wave, 5 Hz 
# Create a cosine fonction 
c = np.cos(2 * np.pi * t)
# Create the first subplot
fig, ax1 = plt.subplots()
# Plot the sine fonction
ax1.plot (t, s, color='blue', label='Sine fonction')
# Add title, labels and grid
ax1.set_title ('Sinusoidal functions')
ax1.set_xlabel('Time (s)')
ax1.set_ylabel('Amplitude')
ax1.grid(True)
# Create a second y axis
ax2 = ax1.twinx()
# plot the cosine fonction
ax2.plot(t, c, color='orange',label='Cosine function')
# Add legends
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')
# Save the plot as a PDF file
plt.savefig('sinusoidal_functions.pdf')
plt.show()
```

```python
# Create a sine fonction 

# Create a function called sinusoid that : takes as input all parameters that you need to define the sinusoid returns a numpy array called x
def sinusoid(time, frequency, amplitude=1, phase=0): # values : amplitude of a sine fonction, phase : no horizontal displacement, define fonction : https://docs.python.org/fr/3.12/tutorial/controlflow.html?utm_source=chatgpt.com#defining-functions
    """Generate a sinusoidal signal.""" #doctring : explain what the fonction do 
    x = amplitude * np.sin(2 * np.pi * frequency * time + phase)
    return x

time = np.linspace(0, 1, 1000) # from :https://numpy.org/doc/stable/reference/generated/numpy.linspace.html?utm_source=chatgpt.com
# Create 2 numpy arrays
s1 = sinusoid(time, frequency=2) # call the fonction with the array
s2 = sinusoid(time, frequency=4)
# plot s1 and s2
plt.plot(time, s1, color='blue', label="Sinus 1")
plt.plot(time, s2, color='orange', label="Sinus 2")

plt.title('Sinusoidal function')
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)
plt.show()

```
